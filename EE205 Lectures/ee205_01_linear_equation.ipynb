# 1. Reduced row echelon form

- Textbook illustration: Use `Sympy tool` and it displays fractions
- In practice, for large matrix, use `LU` decomposition in `NumPy`.

```python
import sympy as sp

# Create a rectangular matrix (3x4)
matrix = sp.Matrix([
    [1, 2, -1, -4],
    [2, 3, -1, -11],
    [-2, 0, -3, 22]
])

# Compute Reduced Row Echelon Form
# rref() returns a tuple: (the matrix, indices of pivot columns)
ref_matrix, pivots = matrix.rref()

print("A is : ")
sp.pprint(matrix)
print("REF Matrix:")
sp.pprint(ref_matrix)
print(f"Pivot columns: {pivots}")
```

```python
import numpy as np
from scipy.linalg import lu

# Define a rectangular matrix
A = np.array([
    [1, 2, -1],
    [2, 3, -1],
    [4, 5, -1],
    [1, 0, 1]
])

# P is permutation matrix, L is lower triangular, U is upper triangular (REF)
P, L, U = lu(A)

# U represents the Row Echelon Form
print("Row Echelon Form (U):")
print(U)
```

## 1.1 Row echelon form of [A y]

- Augment $A$ and $y$ (stack in columns)
- Use `rref` to check if the system $Ax = y$ has a solution
(This is to illustrate only, not used for large system).

```python
import sympy as sp

# 1. Define the coefficient matrix A and the constants vector y
A = sp.Matrix([
    [1, 2, -1],
    [3, 6, -3],
    [-1, 1, 1]
])

y = sp.Matrix([6, 15, 25])

# 2. Create the Augmented Matrix [A | y]
# row_join appends y as a new column to the right of A
augmented_matrix = A.row_join(y)

print("Original Augmented Matrix [A|y]:")
sp.pprint(augmented_matrix) # print in matrix form

# 3. Perform RREF
# rref() returns a tuple: (the_matrix, pivot_column_indices)
rref_matrix, pivots = augmented_matrix.rref()

print("\nMatrix after RREF:")
sp.pprint(rref_matrix)
```

# 2. Solving linear equation $ Ax = y $ when a solution exists.

- Square system: Use `np.linalg.solve(A,y)`
- Fat and tall systems: Use `np.linalg.lstsq(A,y)` or pseudo inverse `pinv`. But you must understand how to interpet the results.

Least-square formulation: minimize $\Vert Ax - y \Vert_2^2$


## 2.1 square system

```python
import numpy as np

print("Square system")
# 3 equations, 3 unknowns
A = np.array([[3, 1, 1],
              [1, 2, 1],
              [1, 1, 2]])
y = np.array([9, 8, 9])

# Use solve() for square, non-singular matrices
x = np.linalg.solve(A, y)

print("The solution is ", x)
print(f"A*x = {A@x}")

print(f"A*x (broadcast) = {A*x}")
```

## 2.2 Fat system

If a solution exists, there are infinitely many solutions. We can find a solution that has specific meaning. For example,

- The one with smallest L2 norm which can be returned by computing pseudo inverse or least-square solution.

Recall that if $z$ is in the nullspace of $A$, and $x$ is a solution to $Ax=y$, then $x+az$ is also a solution for any scalar $a$.

```python
import numpy as np
from scipy.linalg import null_space

# more variables than the number of equations
Afat = np.array([[1, 2, 3,-6,0],
                  [4,3,1,-3,-5],[1,0,-1,2,3]])
yfat = np.array([1,1,4])

# linalg.lstsq (least squares)
# It finds the solution that minimizes ||Ax - y||^2
x1, residuals, rank, s = np.linalg.lstsq(Afat, yfat, rcond=None)

# Using Moore-Penrose Pseudoinverse
Apinv = np.linalg.pinv(Afat)
x2 = Apinv@yfat

print("Underdetermined Solution from least square:", x1)
print("Underdetermined Solution from pseudo inverse:", x2)
print(f"A*x = {Afat@x1}")

# Get the orthonormal basis for the nullspace
ns = null_space(Afat)
print("Vectors in the nullspace of A: \n",ns)
# These are other solutions
x3 = x1 + 2*ns[:,0]
x4 = x1 -3*ns[:,1]

print(f"Check if x3 is another solution: A*x3 = {Afat@x3}")
print(f"Check if x4 is another solution: A*x4 = {Afat@x4}")
```

## 2.3 Tall system

Recall that $Ax = y$ has a solution, when $y$ is in the column span of $A$.

```python
# 4 equations, 2 unknowns
Atall = np.array([[1, 1],
                   [1, 2],
                   [1, 3],
                   [1, 4]])
# This choice of y doesn't make Ax = y solvable
ytall = np.array([6, 11, 14, 19])

# This choice of y makes Ax=y have a solution
# ytall = -2*Atall[:,0]+3*Atall[:,1]

# lstsq is the robust choice here
x1, residuals, rank, s = np.linalg.lstsq(Atall, ytall, rcond=None)

# Using Moore-Penrose Pseudoinverse
Apinv = np.linalg.pinv(Atall)
x2 = Apinv@ytall

print(f"y is {ytall}")
print("Overdetermined (Least Squares) Solution:", x1)
print("Overdetermined (PINV) Solution:", x2)
print(f"A*x = {Atall@x1}")
```

# 3. What happen when the linear system is not solvable?

Recall in theory: $Ax = y$ has no solution iff rank[A y] is not equal to rank[A]

```python
import numpy as np
import sympy as sp

# Square case
print("Square system")
A = np.array([[2,0,-3],[1,1,-4],[4,0,-6]])
A1 = sp.Matrix(A)

# Test with two choices of y (comment on one choice)
# y = np.array([2,4,-9])
y = A[:,0]-2*A[:,1]

y1 = sp.Matrix(y)

# Check what happen
# x = np.linalg.solve(A,y)


# find RREF of [A1 | y1]
augmented_matrix = A1.row_join(y1)
ref_matrix = augmented_matrix.rref()

print("REF matrix is")
sp.pprint(ref_matrix)
# Use SymPy Gaussian Jordan elimination
x1, free_vars = A1.gauss_jordan_solve(y1)

print("A and y")
sp.pprint(A1)
sp.pprint(y1)
print("General Solution:")
sp.pprint(x1)
```

# 4. Homogenous systems

Solve $Ax = 0$.
Important questions:
1. Does the system have non-trivial solutions?
2. What is nullspace of $A$?
3. If the homogenous system have non-trivial solution, parametrize the generation solutions.

```python
import sympy as sp
import numpy as np
from scipy.linalg import null_space

A = np.array([[1,-1,2,-1],[2,1,-2,-2],[-1,2,-4,1],[3,0,0,-3]])
m = np.shape(A)[0]

A1 = sp.Matrix(A)
y1 = sp.zeros(m,1)
x1, free_vars = A1.gauss_jordan_solve(y1)
nullA = A1.nullspace()
nullA2 = null_space(A) # Using Scipy


print("A is \n")
sp.pprint(A1)
print("A general solution is")
sp.pprint(x1)
print("The nullspace of A is \n")
sp.pprint(nullA)

print("Find nullspace using Scipy via SVD")
print(nullA2)
print(f"Verify if it's really in the nullspace \n {A@nullA2}")
```

# Least-squares problem

minimize $\Vert Ax - y \Vert_2^2$ -----(LS)

where $A \in \mathbf{R}^{m \times n}, y \in \mathbf{R}^m$ are problem parameters and the variable is $x \in \mathbf{R}^n$. The LS problem often assumes that $A$ is tall.

**Main result** $x = (A^TA)^{-1}A^T y = A^\dagger y$ (the left inverse of $A$)

we use `np.linalg.lstsq` for linear least square.

Note that another option is `scipy.linalg.lstsq` and `scipy.optimize.least_squares` and the latter is more powerful as it takes nonlinear least-squares as well (more advanced topic).

- When $A$ is tall, solving $Ax=y$ may not have a solution, so we minimize the residual error instead.
- When $A$ is fat, if $Ax=y$ has a solution, you can find infinitely many other solutions (please review), so solving (LS) using returns the least-norm solution.
- When $A$ is square and invertible, solving (LS) should return the solution $x= A^{-1}b$, same as solving linear equation, by returning the zero residual.

# Least-norm problem

minimize $\Vert x \Vert_2$ subject to $Ax = y$

where $A \in \mathbf{R}^{m \times n}, y \in \mathbf{R}^m$ are problem parameters and the variable is $x \in \mathbf{R}^n$. The least-norm problem often assumes that $A$ is fat.

- When $A$ is square and invertible, then $Ax = y$ has a unique solution $A^{-1}y$, so we don't have to minimize its 2 norm, because there's only one solution to choose from.
- When $A$ is tall, $Ax=y$ may or may not have a solution. The vector $y$ must be in the range of $A$ so that $Ax=y$ has a solution. This does not occur easily in real-world data.
- When $A$ is fat, if $Ax=y$ has a solution, it has finitely many solutions, so it makes sense to find **the one** that has the minimum norm.

**Main result**  $x = A^T(AA^T)^{-1}y = A^\dagger y$ (the right inverse of $A$)

How to solve numerically ?

`np.linalg.lstsq` (even though intended for tall system) the command also returns the least-norm solution.

```python
import numpy as np

m = 5
n = 10

A = np.random.randint(-8,8,(m,n))
y = np.random.randint(-8,8,m)

# return x, sum squared residuals, rank of A, singular values of A
x, sumsq_err,rankA,s = np.linalg.lstsq(A,y)
v = np.linalg.pinv(A)@y
print(f"Least-norm solution is {np.round(x,2)}")
print(f"Least-norm solution using pseudo inverse {np.round(v,2)}")
print(f"Minimized 2-norm squared of error is {sumsq_err}") # should be zero
print(f"2-norm of x is {np.linalg.norm(x)}")

# --------- Fill your code ------------------
# just create another u that is a solution to Au = y. This is not the same solutoin as x,v
# u may not have the least-norm

# u = ### find a way to code it.

# print(f"2-norm of u is {np.linalg.norm(u)}")
```
